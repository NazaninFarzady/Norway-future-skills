Section 3 — Data Collection Lessons

## API data collection

An API allows programs to request data from another service.

Important concepts:
- Endpoint: where data is requested from
- Request: asking the API for data
- Response: returned information
- JSON: common format for API data
- authentication
- pagination

Example:
response = requests.get(url): sends request

data = response.json(): converts response into Python objects

API responses are often not ready for analysis
Data extraction requires understanding the JSON structure

Section 4 — Data Cleaning Lessons

Real-world data is messy

## Data cleaning challenges

Real datasets often contain:

- missing values
- inconsistent formats
- nested structures
- incomplete classifications

Cleaning is not only removing errors.
It is also understanding what missing information represents.

## Nested JSON

{
 "occupationList": [
    {
      "name": "Software developer"
    }
 ]
}

Real-world datasets rarely have clean tables. Data extraction requires understanding the structure first.

## Missing values

The occupation example:
Not classified: 8104
Classified: 1896

Missing values are not always errors.

They can represent:
- missing information
- unavailable classification
- employer not providing details

The occupation field had many missing values.

Possible explanations:

- NAV did not classify the job
- Employer did not provide enough information
- Some jobs do not map clearly to occupation categories

Decision:

Keep the missing values instead of removing these jobs.

Reason:

Missing information can itself describe limitations in the dataset.

Section 6 — Skill Analysis Lessons 

## Transforming Data for Analysis

## Skill Analysis Lessons

The purpose of skill analysis was to understand:

- Which skills are most demanded?
- How many skills are usually required per job?
- Do different regions require different skills?

To answer these questions, job-level data was transformed into skill-level data.

### Long format data

From job-level data to skill-level data

Original:

job_id | skills
001    | Python, SQL, Excel

Problem:

A list inside a column is difficult to analyze.

Converted/ Transformation:

job_id | skill
001    | Python
001    | SQL
001    | Excel

Long-format data is easier for:

- counting
- grouping
- visualization
- comparison

## Skill Demand Analysis
### Frequency analysis

Which skills appear most often?

Method:

skill.value_counts()

Frequency counts help identify dominant patterns in a dataset.

### Skill intensity

How many skills are usually required per job?
Method:

groupby("job_id").size()

Aggregation allows us to move from individual records to business-level insights.

## Comparing Groups Using Normalization
### Normalization

Problem:

Oslo has many more jobs than smaller cities.

Raw counts can be misleading.

Raw count:

Oslo:
Digitalisering = 87

But Oslo is large.

So Instead of only counts:

skill_share =
skill count / city total skills

Percentages/ Normalization allows fair comparison between groups of different sizes.

## Comparing groups using normalization

Raw counts can be misleading when groups have different sizes.

Example:

Oslo has more jobs than smaller cities, so it naturally has more skill occurrences.

Instead of only using counts:

skill_share = skill demand / total city skill demand

Normalization allows fair comparison between regions.

Section 7 — Problems and Decisions Log

## Problem: Occupation extraction was mostly empty

Observation:
Only 19% of jobs had occupation classification.

Decision:
Keep the data and treat missing occupation as information.

Reason:
Missing classification may reflect real labor market conditions.

## Challenge: Messy occupation data

Problem:
Occupation information was incomplete.

Decision:
Do not remove records.

Reason:
The dataset should represent the real labor market, including incomplete classification.


## Challenge: Too many unique skills

Problem:
Almost 11,000 unique skills existed.

Decision:
Analyze frequency first, categorize later.

Reason:
Premature grouping may hide important patterns.

## Problem: Over-cleaning data: not everything should be cleaned

Observation:

Some missing or messy values represent real limitations of the source dataset.

Decision:

Do not force all data into a perfect structure.

Reason:

A good analysis should preserve reality, not hide imperfections.
